# 2. Conditional statements: making decisions

A conditional lets a program choose what to do.

```python
if condition:
    # runs when the condition is True
else:
    # runs when the condition is False
```

Python uses **indentation** to identify the block belonging to each condition.

## Real-world example: flagging an air-quality reading

Imagine a simplified sensor rule:

- below 35 → `"normal"`
- 35 to 54 → `"elevated"`
- 55 or above → `"high"`

````python
pm25 = 41

if pm25 < 35:
    status = "normal"
elif pm25 < 55:
    status = "elevated"
else:
    status = "high"

print("PM2.5:", pm25)
print("Status:", status)
````

### What happened, step by step

1. Python checks `pm25 < 35`.
2. If it is false, Python moves to the `elif`.
3. It checks `pm25 < 55`.
4. The first true branch runs.
5. Later branches are skipped.

This pattern is useful in data science when converting a numeric value into a category.

# 3. Combining Boolean expressions

Data-quality rules often involve more than one condition.

- `A and B` is true only when **both** are true.
- `A or B` is true when **at least one** is true.
- `not A` reverses a Boolean value.

## Example: dashboard access

# 4. Lists: storing a sequence of observations

```python

```

Key ideas:

- indexing starts at `0`;
- negative indexes count from the end;
- slicing selects a range;
- lists are mutable, so elements can be changed;

````python
daily_visitors = [124, 141, 138, 155, 149, 170, 162]

print("Whole list:", daily_visitors)
print("First value:", daily_visitors[0])
print("Last value:", daily_visitors[-1])
print("First three:", daily_visitors[:3])
print("Number of observations:", len(daily_visitors))
````

# 5. `while` loops: repeat until a condition changes

A `while` loop is useful when the number of repetitions is not known in advance.

```python

```

A common source of an **infinite loop** is forgetting to update something that affects the condition.

````python
value = 1

while value < 20:
    print(value)
    value = value * 2
````

# 6. `for` loops: process each item in a collection

````python
daily_sales = [120, 85, 140, 63, 170]

for sale in daily_sales:
    if sale >= 100:
        print(sale, "target met")
    else:
        print(sale, "below target")
````
